```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

%matplotlib inline
```

```python
# Load data
# Load the Cars93 dataset
df = pd.read_csv("Cars93.csv")

# Display first few rows
df.head()
```

### 1. Boxplot

```python
# Boxplot for revs per mile
manufacturers = ["Audi", "Hyundai", "Suzuki", "Toyota"]

box_df = df[df["Manufacturer"].isin(manufacturers)]

plt.figure(figsize=(8,5))

sns.boxplot(
    data=box_df,
    x="Manufacturer",
    y="Rev.per.mile"
)

plt.title("Revs per Mile by Manufacturer")
plt.xlabel("Manufacturer")
plt.ylabel("Revs per Mile")

plt.show()
```

**Boxplot question:**  
Which of these manufacturers has the car with the highest revs per mile?


**Boxplot answer:** Toyota


### 2. Histogram

```python
# Histogram of MPG
plt.figure(figsize=(8,5))

plt.hist(
    df["MPG.city"],
    bins=10,
    alpha=0.6,
    label="City MPG"
)

plt.hist(
    df["MPG.highway"],
    bins=10,
    alpha=0.6,
    label="Highway MPG"
)

plt.title("City vs Highway MPG")
plt.xlabel("Miles per Gallon")
plt.ylabel("Number of Cars")

plt.legend()

plt.show()
```

**Histogram question:**  
Is it generally more fuel efficient to drive in the city or on the highway?

**Histogram answer:** Yes

### 3. Lineplot

```python
# Lineplot (Wheelbase vs turning circle)
line_df = df.sort_values("Wheelbase")

plt.figure(figsize=(8,5))

plt.plot(
    line_df["Wheelbase"],
    line_df["Turn.circle"],
    marker="o",
    linestyle="-"
)

plt.title("Wheelbase vs Turning Circle")
plt.xlabel("Wheelbase")
plt.ylabel("Turning Circle")

plt.show()
```

**Lineplot question:**  
What is this relationship? What happens when the wheelbase gets larger?


**Lineplot answer:**  There is a positive relationship. As the wheelbase gets larger, the turning circle becomes larger.


### 4. Barplot

```python
# Barplot for mean horsepower
mean_hp = df.groupby("Type")["Horsepower"].mean().sort_values()

plt.figure(figsize=(9,5))

mean_hp.plot(kind="bar")

plt.title("Mean Horsepower by Car Type")
plt.xlabel("Car Type")
plt.ylabel("Average Horsepower")

plt.xticks(rotation=45)

plt.show()
```

**Barplot question:**  
Does a larger car mean more horsepower? 



**Barplot answer:** Yes
