## Task

In this task you will clean the country column and parse the date column in the **store_income_data_task.csv** file.

```python
# Load up store_income_data.csv
import pandas as pd
from datetime import datetime, date

df = pd.read_csv("store_income_data_task.csv")

df.head()
```

1. Take a look at all the unique values in the "country" column. Then, convert the column to lowercase and remove any trailing white spaces.

```python
df["country"].unique()

df["country"] = (
    df["country"]
    .fillna("")
    .str.lower()
    .str.strip()
)
```

2. Note that there should only be three separate countries. Eliminate all variations, so that 'South Africa', 'United Kingdom' and 'United States' are the only three countries.

```python
country_map = {
    "united states": "United States",
    "united states/": "United States",
    "america": "United States",

    "united kingdom": "United Kingdom",
    "uk": "United Kingdom",
    "u.k.": "United Kingdom",
    "u.k": "United Kingdom",
    "u.k/": "United Kingdom",
    "u.k./": "United Kingdom",
    "britain": "United Kingdom",
    "britain/": "United Kingdom",
    "england": "United Kingdom",

    "sa": "South Africa",
    "sa/": "South Africa",
    "s.a": "South Africa",
    "s.a.": "South Africa",
    "s.a./": "South Africa",
}

# Remove any trailing slashes first
df["country"] = df["country"].str.rstrip("/")

# Replace using the dictionary
df["country"] = df["country"].replace(country_map)
```

3. Create a new column called `days_ago` in the DataFrame that is a copy of the 'date_measured' column but instead it is a number that shows how many days ago it was measured from the current date. Note that the current date can be obtained using `datetime.date.today()`.

```python
# Convert the dates into datetime format
df["date_measured"] = pd.to_datetime(
    df["date_measured"],
    format="%d-%m-%Y"
)

# Calculate the number of days from today
today = pd.Timestamp(date.today())

df["days_ago"] = (today - df["date_measured"]).dt.days

df.head()
```
